## Challenge 6: Copy All Excel Files

Create a script that copies all .xlsx files from an "input" folder to a "backup" folder.
Create the "backup" folder if it doesn't exist.

> Sem pastas criadas: o desafio e criar input/ e backup/.


```python
import shutil
from loguru import logger

def move_files_to_backup_folder(origin_folder, backup_folder):
    """"
        If the backup folder does not exist, create it. Then move the specified files to the backup folder.
    """
    if not os.path.exists(backup_folder):
        os.makedirs(backup_folder)
    for file in os.listdir(origin_folder):
        shutil.move(os.path.join(origin_folder, file), os.path.join(backup_folder, file))
    logger.success("Files moved to backup folder successfully.")
    

move_files_to_backup_folder("input", "backup")       
```

## Challenge 9: Move Old Files

Create a script that moves all .xlsx files older than 7 days from an "input" folder to an "archive" folder.
Use file modification date to determine age.

> Sem pastas criadas: o desafio e criar input/ e archive/.


```python
# Create Archieve folder
os.makedirs("archive/", exist_ok=True)

# Create Files
# with open("archive/teste_3.txt", "w") as q:
#     pass

def move_files_before_x_days(primary_folder, destiny_folder, days):
    now= datetime.now() #2026, 11, 12, 15, 3, 40, 537656
    for i in os.listdir(primary_folder):
        full_path= os.path.join(primary_folder, i)
        file_time = os.path.getmtime(full_path)
        # Convert to float
        file_time = datetime.fromtimestamp(file_time)

        # Result
        days_left= now - file_time
        #print(f"================ File: {i} ================\nDays: {days_left.days}\nMinutes: {days_left.min}\nSeconds: {days_left.seconds}\n")

        if int(days_left.days) >= days:
            shutil.move(full_path, destiny_folder)
            logger.success(f"File {i} Moved to Folder: {destiny_folder}\n ({days_left.days}) Days ")
        else:
            logger.info(f"File {i} doesn't have your required amount of days")



move_files_before_x_days(primary_folder="input",
                         destiny_folder = "archive",
                         days=1)
```

## Challenge 10: Check Empty File

Create a function that checks if an Excel file is empty (no data rows) and returns True or False.
Consider that headers don't count as data.

> Apoio: D10_EmptyFile.xlsx (so cabecalhos).


```python
def empty_file(file_path):
    """Input the file and it will return if the file is empty ("TRUE") or not"""
    try:
        df = pd.read_excel(file_path)
        if df.empty:
            return True
        else:
            return False
    except Exception as e:
        print(f"Error to read file: {e}")



empty_file("D10_EmptyFile.xlsx") # TRUE
empty_file("D07_Clients.xlsx") # FALSE
```

## Challenge 12: Filter Data by Condition

Read an Excel file and filter all rows where the "Status" column equals "Pending".
Save the filtered result to a new file.
Create a parameterized function that accepts column name and value as arguments.

> Apoio: D07_Clients.xlsx


```python
def filter_by_condition_status(filepath, filter, output_filepath):
    try:
        data = pd.read_excel(filepath)
        filtered_data = data[data['Status'].str.lower() == filter.lower()]
        filtered_data.to_excel(output_filepath, index=False)
        logger.success(f"Filtered succesfully, please check into: {output_filepath}")  
    except Exception as e:
        logger.error(f"Error in filter_by_condition_status function:\n{e}")


filter_by_condition_status("D07_Clients.xlsx", "PENDING", "Pending_D07_Clients.xlsx")
```

## Challenge 13: Batch Processing with Logging

Create a script that processes all .xlsx files in a folder, reads the "Value" column, and calculates the total sum.
Log each file: filename, row count, and sum.
Handle corrupted files without breaking the loop.

> Apoio: D11_sales_*.xlsx e D13_corrupted.xlsx (corrompido de proposito).


```python
def sum_value_from_folder(folder):
    files= os.listdir(folder)
    total_sum=0

    for i in files:
        if i.endswith(".xlsx"):
            try:
                data = pd.read_excel(f"{folder}\{i}")
                file_sum = data['Value'].sum()
                total_sum += file_sum

                logger.info(f"File: {i} | Rows: {len(data)} | Sum: {file_sum}")
            except Exception as e:
                logger.error(f"Corrupted file: {i},\nError: {e}")
                # Move for corrupted folder
                os.makedirs("Corrupted", exist_ok=True)
                shutil.move(f"{folder}\{i}", "Corrupted")
        else:
            pass
    return total_sum

grand_total = sum_value_from_folder("files")
logger.success(f"Grand Total of all files: {grand_total}")
```
